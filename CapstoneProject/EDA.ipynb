## Question 1: Are AI Tools Associated With Better Worker Efficiency?

```python
productivity_corr_cols = [
    'productivity_score',
    'ai_tool_usage_hours_per_week',
    'tasks_automated_percent',
    'manual_work_hours_per_week',
    'error_rate_percent',
    'focus_hours_per_day',
    'work_life_balance_score',
    'burnout_risk_score',
    'learning_time_hours_per_week',
    'meeting_hours_per_week',
    'collaboration_hours_per_week',
    'task_complexity_score'
]

productivity_correlations = (
    productivity_df[productivity_corr_cols]
    .corr(numeric_only=True)['productivity_score']
    .sort_values(ascending=False)
    .to_frame('corr_with_productivity_score')
)

productivity_correlations
```

```python
productivity_by_role = (
    productivity_df
    .groupby('job_role')
    .agg(
        employees=('Employee_ID', 'count'),
        avg_productivity=('productivity_score', 'mean'),
        avg_ai_hours=('ai_tool_usage_hours_per_week', 'mean'),
        avg_tasks_automated=('tasks_automated_percent', 'mean'),
        avg_manual_hours=('manual_work_hours_per_week', 'mean'),
        avg_error_rate=('error_rate_percent', 'mean'),
        avg_burnout_score=('burnout_risk_score', 'mean')
    )
    .sort_values('avg_productivity', ascending=False)
    .round(2)
)

productivity_by_role
```

## Question 2: Is AI Adoption Affecting Burnout or Stress?


```python
burnout_corr_cols = [
    'burnout_risk_score',
    'productivity_score',
    'ai_tool_usage_hours_per_week',
    'tasks_automated_percent',
    'manual_work_hours_per_week',
    'work_life_balance_score',
    'focus_hours_per_day',
    'meeting_hours_per_week',
    'deadline_pressure_level'
]

burnout_correlations = (
    productivity_df[burnout_corr_cols]
    .corr(numeric_only=True)['burnout_risk_score']
    .sort_values(ascending=False)
    .to_frame('corr_with_burnout_risk_score')
)

burnout_correlations
```

```python
worker_ai_burnout_summary = (
    burnout_df
    .groupby(['ai_adoption_stage', 'fear_of_ai_replacement'])
    .agg(
        employees=('employee_id', 'count'),
        avg_ai_hours_daily=('hours_with_ai_assistance_daily', 'mean'),
        avg_productivity=('productivity_score', 'mean'),
        avg_burnout=('burnout_score', 'mean'),
        avg_job_satisfaction=('job_satisfaction_1_5', 'mean'),
        avg_task_replacement_pct=('ai_replaces_my_tasks_pct', 'mean')
    )
    .sort_values(['avg_burnout', 'avg_task_replacement_pct'], ascending=False)
    .round(2)
)

worker_ai_burnout_summary
```

```python
attrition_burnout_summary = (
    burnout_df
    .groupby('attrition_risk')
    .agg(
        employees=('employee_id', 'count'),
        avg_burnout=('burnout_score', 'mean'),
        avg_productivity=('productivity_score', 'mean'),
        avg_job_satisfaction=('job_satisfaction_1_5', 'mean'),
        avg_ai_hours_daily=('hours_with_ai_assistance_daily', 'mean'),
        avg_task_replacement_pct=('ai_replaces_my_tasks_pct', 'mean')
    )
    .sort_values('avg_burnout', ascending=False)
    .round(2)
)

attrition_burnout_summary
```

## Question 3: Is There an Optimal Balance Between AI Use and Manual Coding/Work Time?


```python
productivity_df = productivity_df.copy()
productivity_df['ai_usage_bucket'] = pd.qcut(
    productivity_df['ai_tool_usage_hours_per_week'],
    q=5,
    labels=['Very low', 'Low', 'Medium', 'High', 'Very high']
)
productivity_df['ai_to_manual_ratio'] = (
    productivity_df['ai_tool_usage_hours_per_week'] /
    productivity_df['manual_work_hours_per_week'].replace(0, np.nan)
)
productivity_df['ai_manual_balance_bucket'] = pd.qcut(
    productivity_df['ai_to_manual_ratio'],
    q=5,
    labels=['Mostly manual', 'Manual leaning', 'Balanced', 'AI leaning', 'AI heavy']
)

ai_usage_outcomes = (
    productivity_df
    .groupby('ai_usage_bucket', observed=True)
    .agg(
        employees=('Employee_ID', 'count'),
        avg_ai_hours=('ai_tool_usage_hours_per_week', 'mean'),
        avg_manual_hours=('manual_work_hours_per_week', 'mean'),
        avg_productivity=('productivity_score', 'mean'),
        avg_burnout_score=('burnout_risk_score', 'mean'),
        avg_tasks_automated=('tasks_automated_percent', 'mean'),
        avg_error_rate=('error_rate_percent', 'mean'),
        avg_work_life_balance=('work_life_balance_score', 'mean')
    )
    .round(2)
)

ai_usage_outcomes
```

```python
ai_manual_balance_outcomes = (
    productivity_df
    .groupby('ai_manual_balance_bucket', observed=True)
    .agg(
        employees=('Employee_ID', 'count'),
        avg_ai_to_manual_ratio=('ai_to_manual_ratio', 'mean'),
        avg_ai_hours=('ai_tool_usage_hours_per_week', 'mean'),
        avg_manual_hours=('manual_work_hours_per_week', 'mean'),
        avg_productivity=('productivity_score', 'mean'),
        avg_burnout_score=('burnout_risk_score', 'mean'),
        avg_error_rate=('error_rate_percent', 'mean')
    )
    .round(2)
)

ai_manual_balance_outcomes
```

```python
dev_productivity_df = dev_productivity_df.copy()
dev_productivity_df['ai_usage_bucket'] = pd.qcut(
    dev_productivity_df['ai_usage_hours'],
    q=5,
    labels=['Very low', 'Low', 'Medium', 'High', 'Very high']
)

coding_ai_outcomes = (
    dev_productivity_df
    .groupby('ai_usage_bucket', observed=True)
    .agg(
        sessions=('task_success', 'count'),
        avg_ai_usage_hours=('ai_usage_hours', 'mean'),
        success_rate=('task_success', 'mean'),
        avg_hours_coding=('hours_coding', 'mean'),
        avg_commits=('commits', 'mean'),
        avg_bugs_reported=('bugs_reported', 'mean'),
        avg_cognitive_load=('cognitive_load', 'mean'),
        avg_sleep_hours=('sleep_hours', 'mean'),
        avg_distractions=('distractions', 'mean')
    )
    .round(2)
)

coding_ai_outcomes
```

## Question 4: Is AI Adoption Giving Companies an Adequate ROI Signal?

```python
company_roi_cols = [
    'productivity_change_percent',
    'time_saved_per_week',
    'task_automation_rate',
    'ai_adoption_rate',
    'ai_maturity_score',
    'ai_budget_percentage',
    'ai_investment_per_employee',
    'revenue_growth_percent',
    'cost_reduction_percent',
    'innovation_score',
    'employee_satisfaction_score',
    'ai_failure_rate'
]

company_roi_correlations = (
    company_adoption_df[company_roi_cols]
    .corr(numeric_only=True)['productivity_change_percent']
    .sort_values(ascending=False)
    .to_frame('corr_with_productivity_change_percent')
)

company_roi_correlations
```

```python
company_adoption_df = company_adoption_df.copy()
company_adoption_df['ai_maturity_bucket'] = pd.qcut(
    company_adoption_df['ai_maturity_score'],
    q=5,
    labels=['Very low', 'Low', 'Medium', 'High', 'Very high']
)

roi_by_maturity = (
    company_adoption_df
    .groupby('ai_maturity_bucket', observed=True)
    .agg(
        responses=('response_id', 'count'),
        avg_ai_adoption_rate=('ai_adoption_rate', 'mean'),
        avg_ai_budget_pct=('ai_budget_percentage', 'mean'),
        avg_investment_per_employee=('ai_investment_per_employee', 'mean'),
        avg_productivity_change=('productivity_change_percent', 'mean'),
        avg_time_saved=('time_saved_per_week', 'mean'),
        avg_revenue_growth=('revenue_growth_percent', 'mean'),
        avg_cost_reduction=('cost_reduction_percent', 'mean'),
        avg_failure_rate=('ai_failure_rate', 'mean'),
        avg_employee_satisfaction=('employee_satisfaction_score', 'mean')
    )
    .round(2)
)

roi_by_maturity
```

## Question 5: Does AI Help Increase Innovation?


```python
innovation_cols = [
    'innovation_score',
    'ai_adoption_rate',
    'ai_maturity_score',
    'ai_budget_percentage',
    'num_ai_tools_used',
    'ai_projects_active',
    'task_automation_rate',
    'time_saved_per_week',
    'productivity_change_percent',
    'revenue_growth_percent',
    'cost_reduction_percent',
    'ai_failure_rate'
]

innovation_correlations = (
    company_adoption_df[innovation_cols]
    .corr(numeric_only=True)['innovation_score']
    .sort_values(ascending=False)
    .to_frame('corr_with_innovation_score')
)

innovation_correlations
```

```python
innovation_by_use_case = (
    company_adoption_df
    .groupby('ai_use_case')
    .agg(
        responses=('response_id', 'count'),
        avg_innovation_score=('innovation_score', 'mean'),
        avg_productivity_change=('productivity_change_percent', 'mean'),
        avg_revenue_growth=('revenue_growth_percent', 'mean'),
        avg_cost_reduction=('cost_reduction_percent', 'mean'),
        avg_failure_rate=('ai_failure_rate', 'mean')
    )
    .sort_values('avg_innovation_score', ascending=False)
    .round(2)
)

innovation_by_use_case
```

## Question 6: Which Job Roles Feel Most Threatened by AI Replacement?

This uses the worker burnout file because it includes perceived task replacement and fear of AI replacement.


```python
threat_by_role = (
    burnout_df
    .groupby('job_role')
    .agg(
        employees=('employee_id', 'count'),
        avg_task_replacement_pct=('ai_replaces_my_tasks_pct', 'mean'),
        high_fear_share=('fear_of_ai_replacement', lambda s: (s == 'High').mean()),
        avg_burnout=('burnout_score', 'mean'),
        avg_productivity=('productivity_score', 'mean'),
        avg_job_satisfaction=('job_satisfaction_1_5', 'mean'),
        avg_ai_hours_daily=('hours_with_ai_assistance_daily', 'mean')
    )
    .sort_values(['avg_task_replacement_pct', 'high_fear_share'], ascending=False)
    .round(2)
)

threat_by_role
```

## Question 7: Which Industries Benefit Most From AI?

```python
industry_benefit_summary = (
    company_adoption_df
    .groupby('industry')
    .agg(
        responses=('response_id', 'count'),
        avg_ai_adoption_rate=('ai_adoption_rate', 'mean'),
        avg_productivity_change=('productivity_change_percent', 'mean'),
        avg_time_saved=('time_saved_per_week', 'mean'),
        avg_revenue_growth=('revenue_growth_percent', 'mean'),
        avg_cost_reduction=('cost_reduction_percent', 'mean'),
        avg_innovation_score=('innovation_score', 'mean'),
        avg_employee_satisfaction=('employee_satisfaction_score', 'mean'),
        avg_failure_rate=('ai_failure_rate', 'mean'),
        avg_jobs_displaced=('jobs_displaced', 'mean'),
        avg_jobs_created=('jobs_created', 'mean')
    )
    .round(2)
)

# Rank industries across several positive AI-benefit signals.
benefit_rank_cols = [
    'avg_ai_adoption_rate',
    'avg_productivity_change',
    'avg_time_saved',
    'avg_revenue_growth',
    'avg_cost_reduction',
    'avg_innovation_score',
    'avg_employee_satisfaction'
]
industry_benefit_summary['benefit_rank_score'] = (
    industry_benefit_summary[benefit_rank_cols]
    .rank(ascending=False)
    .mean(axis=1)
)

industry_benefit_summary.sort_values('benefit_rank_score')
```

```python
country_readiness_summary = (
    company_enriched_df
    .groupby(['country', 'region', 'country_ai_policy'])
    .agg(
        responses=('response_id', 'count'),
        avg_digital_maturity=('digital_maturity_index', 'mean'),
        avg_internet_penetration=('internet_penetration', 'mean'),
        avg_researchers_per_million=('ai_researchers_per_million', 'mean'),
        avg_ai_adoption_rate=('ai_adoption_rate', 'mean'),
        avg_productivity_change=('productivity_change_percent', 'mean'),
        avg_innovation_score=('innovation_score', 'mean'),
        avg_revenue_growth=('revenue_growth_percent', 'mean')
    )
    .sort_values('avg_productivity_change', ascending=False)
    .round(2)
)

country_readiness_summary.head(15)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 11
```

### AI Tool Usage and Productivity


```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=productivity_df,
    x='ai_tool_usage_hours_per_week',
    y='productivity_score',
    hue='burnout_risk_level',
    alpha=0.65
)
sns.regplot(
    data=productivity_df,
    x='ai_tool_usage_hours_per_week',
    y='productivity_score',
    scatter=False,
    color='black'
)
plt.title('AI Tool Usage vs Productivity Score')
plt.xlabel('AI tool usage hours per week')
plt.ylabel('Productivity score')
plt.legend(title='Burnout risk level', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Productivity and Burnout by AI Usage Bucket


```python
plot_df = ai_usage_outcomes.reset_index().melt(
    id_vars='ai_usage_bucket',
    value_vars=['avg_productivity', 'avg_burnout_score'],
    var_name='metric',
    value_name='average_score'
)
plot_df['metric'] = plot_df['metric'].replace({
    'avg_productivity': 'Productivity score',
    'avg_burnout_score': 'Burnout risk score'
})

plt.figure(figsize=(11, 6))
sns.barplot(
    data=plot_df,
    x='ai_usage_bucket',
    y='average_score',
    hue='metric'
)
plt.title('Average Productivity and Burnout by AI Usage Bucket')
plt.xlabel('AI usage bucket')
plt.ylabel('Average score')
plt.xticks(rotation=25)
plt.legend(title='Metric')
plt.tight_layout()
plt.show()
```

### AI-to-Manual Work Balance


```python
balance_plot_df = ai_manual_balance_outcomes.reset_index().melt(
    id_vars='ai_manual_balance_bucket',
    value_vars=['avg_productivity', 'avg_burnout_score'],
    var_name='metric',
    value_name='average_score'
)
balance_plot_df['metric'] = balance_plot_df['metric'].replace({
    'avg_productivity': 'Productivity score',
    'avg_burnout_score': 'Burnout risk score'
})

plt.figure(figsize=(11, 6))
sns.lineplot(
    data=balance_plot_df,
    x='ai_manual_balance_bucket',
    y='average_score',
    hue='metric',
    marker='o',
    linewidth=2.5
)
plt.title('Productivity and Burnout Across AI-to-Manual Work Balance')
plt.xlabel('AI/manual balance bucket')
plt.ylabel('Average score')
plt.xticks(rotation=25)
plt.legend(title='Metric')
plt.tight_layout()
plt.show()
```

### Company AI Maturity and ROI Signals
